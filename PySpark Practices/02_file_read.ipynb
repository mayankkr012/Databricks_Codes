```python
data = [
    (101, "Rahul", 55000, "IT"),
    (102, "Priya", 62000, "HR"),
    (103, "Amit", 70000, "Finance"),
    (104, "Sneha", 60000, "IT")
]

columns = ["Emp_Id", "Name", "Salary", "Department"]

df = spark.createDataFrame(data, columns)

df.write.mode('overwrite').parquet('/Volumes/workspace/default/practice/output_parquet')
df.display()
df.printSchema()
df.show(2)
df.count()
df.describe().display()
df.columns
```

```python
data = [
    (101, "Rahul", 55000, "IT"),
    (102, "Priya", 62000, "HR"),
    (103, "Amit", 70000, "Finance"),
    (104, "Sneha", 60000, "IT"),
    (105, "Rohit", 58000, "HR")
]

columns = ["Emp_Id", "Name", "Salary", "Department"]

df = spark.createDataFrame(data, columns)

df.write.mode("overwrite") \
    .partitionBy("Department") \
    .parquet("/Volumes/workspace/default/practice/partitioned_parquet")
```
