```python
df.select('Name','Salary').display()
```

```python
df.filter(col('Salary')>60000).display()
```

```python
df.filter((col('Salary')>55000)&(col('Department')=='IT')).display()
```

```python
df.withColumn('Bonus',lit(5000)).display()
```

```python
df.withColumn('Salary_After_Bonus',col('Salary')+10000).display()
```

```python
df.withColumnRenamed('Department','Dept').display()
```

```python
df.orderBy(col('Salary').desc()).display()
```
