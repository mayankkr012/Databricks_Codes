```python
data = [
    (101, "Rahul", 55000, "IT"),
    (102, "Priya", 62000, "HR"),
    (103, "Rahul", 55000, "IT"),
    (104, "Sneha", 70000, "Finance"),
    (105, "Priya", 62000, "HR")
]

columns = ["Emp_Id", "Name", "Salary", "Department"]

df_dup = spark.createDataFrame(data, columns)

df_dup.show()
```

```python
df_dup.dropDuplicates().show()
```

```python
df_dup.dropDuplicates(['Department']).show()
```

```python
df_dup.groupBy('Department').agg(sum('Salary'),avg('Salary')).show()
```

```python
df_dup.groupBy('Department').agg(sum('Salary').alias('Total_Salary'),avg('Salary').alias('Avg_Salary')).show()
```

```python
from pyspark.sql.functions import count
df_dup.groupBy('Department').agg(count('*')).show()
```

```python
df_dup.groupBy('Department').agg(count('*').alias('Employee_Count'),sum('Salary').alias('Total_Salary'),avg('Salary').alias('Avg_Salary')).show()
```

```python
df_dup.groupBy('Department').agg(count('*').alias('Employee_Count'),sum('Salary').alias('Total_Salary'),avg('Salary').alias('Avg_Salary'),min('Salary').alias('Min_Salary'),max('Salary').alias('Max_Salary')).show()
```

```python
from pyspark.sql.functions import when
df.withColumn('Salary_Category',when(col('Salary')>55000,'High').otherwise('Low')).show()
```

```python
df.withColumn('Salary_Level',when(col('Salary')>=65000,'High').when(col('Salary')>=60000,'Medium').otherwise('Low')).show()
```
