```python
from pyspark.sql.functions import current_date
df.withColumn('Today',current_date()).display()
```

```python
from pyspark.sql.functions import current_timestamp
df.withColumn('Current_Time',current_timestamp()).display()
```

```python
from pyspark.sql.functions import current_date,date_add,date_sub
df.withColumn('Future_Date',date_add(current_date(),10)).display()
```

```python
df.withColumn("Past_Date",date_sub(current_date(),15)).display()
```

```python
from pyspark.sql.functions import year
df.withColumn('Year',year(current_date())).display()
```

```python
from pyspark.sql.functions import year,month,dayofmonth
df.withColumn('Current_Month',month(current_date())).withColumn('Current_Day',dayofmonth(current_date())).display()
```

```python
from pyspark.sql.functions import datediff
df.withColumn('Days_Difference',datediff(current_date(),date_sub(current_date(),7))).display()
```

```python
data = [
    (1, 'rahul'),
    (2, 'PRIYA'),
    (3, 'aMiT')
]

columns = ['Id', 'Name']

df_str = spark.createDataFrame(data, columns)
df_str.display()
```

```python
from pyspark.sql.functions import upper
df_str.withColumn('Upper_Name',upper('Name')).display()
```

```python
from pyspark.sql.functions import lower
df_str.withColumn('Lower_Name',lower('Name')).display()
```

```python
from pyspark.sql.functions import initcap
df_str.withColumn('Proper_Name',initcap('Name')).show()
```

```python
from pyspark.sql.functions import length
df_str.withColumn('Name_Length',length('Name')).display()
```
