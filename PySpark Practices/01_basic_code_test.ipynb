```python
data=[
    (1,'mayank',25),
    (2,'sachin',26),
    (3,'saurav',27)
]
columns=['id','name','age']
df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.printSchema()
```

```python
df.select('name').show()
```

```python
df.select('name','age').show()
```

```python
df.filter('age>26').show()
```

```python

df.where('age>26').show()
```

```python
from pyspark.sql.functions import col
df2=df.withColumn('age_after_5_years',col('age')+5)
df2.show()
```

```python
df3=df.withColumnRenamed('name','Full_Name')
df3.show()
```

```python
df.drop('age').show()
```

```python
df.groupby('age').count().show()
```

```python
df.orderBy('age').show()
```

```python
df.orderBy(df.age.desc()).show()
```

```python
data=[
    (1,'Mayank',15000,'Esim'),
    (2,'Vijaya',17000,'Wcom'),
    (3,'Shanti',20000,'Quality'),
    (4,'Shagun',16000,'DND')
]
columns=['Id','Name','Salary','Department']
df=spark.createDataFrame(data,columns)
df.show()
```
